```python

# STEP 2: MISSING VALUES ANALYSIS

# 1. Count missing values
missing_count = df.isnull().sum()

print("MISSING VALUES COUNT:")
print(missing_count)


# 2. Calculate missing percentage
missing_percentage = (missing_count / len(df)) * 100

print("\nMISSING VALUE PERCENTAGE:")
print(missing_percentage)


# 3. Display columns with missing values
missing_data = missing_percentage[missing_percentage > 0]

print("\nCOLUMNS WITH MISSING VALUES:")
print(missing_data)


# 4. Create final missing values report
missing_report = pd.DataFrame({
    "Missing Count": missing_count,
    "Missing Percentage": missing_percentage
})

missing_report = missing_report[
    missing_report["Missing Count"] > 0
]

print("\nFINAL MISSING VALUES REPORT:")
print(missing_report)
```

```python

# Analyse Income column

print("INCOME STATISTICS")

print("Mean:", df["Income"].mean())
print("Median:", df["Income"].median())
print("Mode:", df["Income"].mode()[0])
print("Minimum:", df["Income"].min())
print("Maximum:", df["Income"].max())

print("\nIncome Description:")
print(df["Income"].describe())
```

```python

import matplotlib.pyplot as plt

plt.figure(figsize=(10, 5))

plt.hist(df["Income"].dropna(), bins=30)

plt.title("Customer Income Distribution")
plt.xlabel("Income")
plt.ylabel("Number of Customers")

plt.show()
```

```python
# Compare mean and median

print("Mean Income:", df["Income"].mean())
print("Median Income:", df["Income"].median())

# Check extreme values
print("Maximum Income:", df["Income"].max())
print("Minimum Income:", df["Income"].min())
```

```python
# Create a separate cleaned dataset
df_clean = df.copy()

# Fill missing Income values using median
median_income = df_clean["Income"].median()

df_clean["Income"] = df_clean["Income"].fillna(median_income)

# Verify missing values
print("Missing Income values:", df_clean["Income"].isnull().sum())
```

```python

# Check missing values in the entire cleaned dataset

print("Missing values before cleaning:")
print(df.isnull().sum().sum())

print("\nMissing values after cleaning:")
print(df_clean.isnull().sum().sum())
```

```python

# Check duplicate records

duplicate_count = df_clean.duplicated().sum()

print("Total duplicate records:", duplicate_count)
```

```python
# Check data types of all columns

print(df_clean.dtypes)
```

```python

# Convert Dt_Customer from object to datetime

df_clean["Dt_Customer"] = pd.to_datetime(
    df_clean["Dt_Customer"],
    format="%d-%m-%Y",
    errors="coerce"
)

# Check updated data types
print(df_clean.dtypes)

# Check whether conversion created missing dates
print("Invalid dates:", df_clean["Dt_Customer"].isnull().sum())
```

```python
df_clean.info()
```

```python
print("Duplicate Customer IDs:", df_clean["ID"].duplicated().sum())
```

```python
print(df_clean["Education"].value_counts(dropna=False))
```

```python
print(df_clean["Marital_Status"].value_counts(dropna=False))
```

```python

# Check for leading and trailing spaces

for col in ["Education", "Marital_Status"]:
    print("\nColumn:", col)

    print("Values with extra spaces:")

    mask = df_clean[col].astype("string").ne(
        df_clean[col].astype("string").str.strip()
    )

    print(df_clean.loc[mask, col].unique())
```

```python

# Check birth year statistics

print("Birth Year Statistics:")
print(df_clean["Year_Birth"].describe())

# Check 10 oldest birth years
print("\nOldest birth years:")
print(df_clean["Year_Birth"].sort_values().head(10))

# Check 10 latest birth years
print("\nLatest birth years:")
print(df_clean["Year_Birth"].sort_values(ascending=False).head(10))
```

```python

# Identify unusually old birth years

suspicious_birth_years = df_clean[
    df_clean["Year_Birth"] <= 1900
]

print("Suspicious birth year records:")
print(suspicious_birth_years[
    ["ID", "Year_Birth", "Education", "Marital_Status", "Income"]
])

print("\nTotal suspicious records:",
      len(suspicious_birth_years))

```

```python

# Flag suspicious birth years

df_clean["Birth_Year_Flag"] = np.where(
    df_clean["Year_Birth"] <= 1900,
    "Review Required",
    "Valid"
)

# Check flagged records

print(
    df_clean[df_clean["Birth_Year_Flag"] == "Review Required"][
        ["ID", "Year_Birth", "Birth_Year_Flag"]
    ]
)

# Count flagged records
print("\nFlagged records:",
      (df_clean["Birth_Year_Flag"] == "Review Required").sum())
```

```python

# FINAL DATA QUALITY CHECK

print("Total Rows:", len(df_clean))
print("Total Columns:", len(df_clean.columns))

print("\nMissing Values:")
print(df_clean.isnull().sum().sum())

print("\nDuplicate Rows:")
print(df_clean.duplicated().sum())

print("\nDuplicate Customer IDs:")
print(df_clean["ID"].duplicated().sum())

print("\nDate Data Type:")
print(df_clean["Dt_Customer"].dtype)

print("\nBirth Year Anomalies:")
print((df_clean["Birth_Year_Flag"] == "Review Required").sum())

```

```python
df_clean.to_csv("cleaned_customer_data.csv", index=False)

print("Cleaned dataset saved successfully!")
```
